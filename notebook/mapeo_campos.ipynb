```python
import pandas as pd
```

```python
df = pd.read_csv("../csv/used_cars.csv")
df2 = pd.read_csv("../csv/train.csv")

df = pd.concat([df, df2])
```

```python
df.head()
```

```python
columnas = ['model', 'model_year', 'milage', 'fuel_type', 'engine', 'transmission', 'ext_col', 'int_col', 'accident', 'clean_title', 'price', 'id']
map_model_df = df.drop(columnas, axis=1)
map_model_df['brand_id'] = map_model_df['brand'].astype('category').cat.codes + 1
```

```python
map_model_df.to_csv("..\\csv\\mapeo_brand.csv", index=False)
```
